```python
#EXPERIENT-11#
# Top-Down Parser for Grammar:
# S -> aAb
# A -> c

string = input("NLP is a good language ")
pos = 0

def match(ch):
    global pos
    if pos < len(string) and string[pos] == ch:
        pos += 1
        return True
    return False

def A():
    return match('c')

def S():
    if match('a'):
        if A():
            if match('b'):
                return True
    return False

if S() and pos == len(string):
    print("String Accepted")
else:
    print("String Rejected")
```

```python
#EXPERIMENT-12#
import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> NP VP
NP -> 'John'
VP -> 'runs'
""")

parser = EarleyChartParser(grammar)

sentence = input("Enter sentence: ").split()

try:
    for tree in parser.parse(sentence):
        print(tree)
except ValueError:
    print("Sentence Rejected")
```

```python
#EXPERIMENT-13#
import nltk
from nltk import CFG
from nltk.parse import ChartParser

grammar = CFG.fromstring("""
S -> NP VP
NP -> 'John'
VP -> V NP
V -> 'likes'
NP -> 'apples'
""")

parser = ChartParser(grammar)

sentence = "John likes apples".split()

for tree in parser.parse(sentence):
    print(tree)
```

```python
#EXPERIMENT-15#
from nltk import PCFG
from nltk.parse import ViterbiParser

grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> 'John' [1.0]
VP -> 'runs' [1.0]
""")

parser = ViterbiParser(grammar)

sentence = "John runs".split()

for tree in parser.parse(sentence):
    print(tree)
```

```python
#EXPERIMENT-20#
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample documents
documents = [
    "Natural Language Processing is interesting",
    "Machine Learning is a branch of Artificial Intelligence",
    "Python is widely used in NLP",
    "Information Retrieval uses TF IDF for document ranking"
]

# User query
query = input("Enter search query: ")

# TF-IDF Vectorization
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

query_vector = vectorizer.transform([query])

# Calculate similarity
similarity = cosine_similarity(query_vector, tfidf_matrix)

print("\nDocument Ranking:")
for i, score in enumerate(similarity[0]):
    print("Document", i + 1, ":", round(score, 3))
```

```python
#EXPERIMENT-21#
import nltk
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet

# Run once if required
# nltk.download('punkt')
# nltk.download('punkt_tab')
# nltk.download('averaged_perceptron_tagger')
# nltk.download('averaged_perceptron_tagger_eng')
# nltk.download('wordnet')

sentence = input("Enter a sentence: ")

words = word_tokenize(sentence)
tags = pos_tag(words)

print("Noun Phrases and Meanings:")

for word, tag in tags:
    if tag.startswith("NN"):
        print("\nNoun:", word)
        syn = wordnet.synsets(word)
        if syn:
            print("Meaning:", syn[0].definition())
        else:
            print("Meaning not found.")
```

```python
#EXPERIMENT-23#
text = input("Enter the text: ").lower()

words = text.split()

common = {}

for word in words:
    common[word] = common.get(word, 0) + 1

count = 0

for value in common.values():
    if value > 1:
        count += 1

if count > 0:
    print("The text is Coherent.")
else:
    print("The text is Less Coherent.")
```
